```python
import os
import json
import numpy as np
import pandas as pd

METADATA_PATH = r"F:\eeg\metadata.csv"

OUTPUT_DIR = r"F:\eeg\transformer_dataset"
os.makedirs(OUTPUT_DIR, exist_ok=True)

metadata = pd.read_csv(METADATA_PATH)

FS = 125
WINDOWS_PER_TRIAL = 156
SAMPLES_PER_WINDOW = 8

# Exclude T8 (EXG Channel 13)
EEG_COLUMNS = [
    f"EXG Channel {i}"
    for i in range(16)
    if i != 13
]

print("Metadata shape:", metadata.shape)
print("Participants:", metadata["participant"].nunique())
print("Trials:", metadata["trial_uid"].nunique())
print("Raw recordings:", metadata["raw_path2"].nunique())
print("Labels:", sorted(metadata["label"].unique()))
```

```python
def read_raw_eeg(path):

    raw = pd.read_csv(
        path,
        skiprows=4,
        low_memory=False
    )

    # Fix leading spaces in column names
    raw.columns = raw.columns.str.strip()

    required = EEG_COLUMNS + ["Sample Index"]

    for col in required:
        raw[col] = pd.to_numeric(
            raw[col],
            errors="coerce"
        )

    if raw[required].isna().any().any():
        raise ValueError(
            f"Missing or invalid EEG/sample-index values: {path}"
        )

    # Verify sequential OpenBCI sample counter.
    # Counter rollover is handled using modulo 256.
    sample_index = raw["Sample Index"].to_numpy()

    differences = np.diff(sample_index) % 256

    if not np.all(differences == 2):
        raise ValueError(
            f"Unexpected sample counter sequence: {path}"
        )

    eeg = raw[EEG_COLUMNS].to_numpy(
        dtype=np.float32
    )

    return eeg
```

```python
def read_raw_eeg(path):

    raw = pd.read_csv(
        path,
        skiprows=4,
        low_memory=False
    )

    raw.columns = raw.columns.str.strip()

    missing = [
        col for col in EEG_COLUMNS
        if col not in raw.columns
    ]

    if missing:
        raise ValueError(
            f"Missing EEG columns in {path}: {missing}"
        )

    # Read only the 15 required EEG channels
    for col in EEG_COLUMNS:
        raw[col] = pd.to_numeric(
            raw[col],
            errors="coerce"
        )

    # Do not silently remove rows: that would change row indices
    if raw[EEG_COLUMNS].isna().any().any():
        raise ValueError(
            f"Missing/non-numeric EEG samples in: {path}"
        )

    eeg = raw[EEG_COLUMNS].to_numpy(
        dtype=np.float32
    )

    return eeg
```

```python

def extract_all_trials(metadata):

    training_trials = []

    for raw_path, file_meta in metadata.groupby(
        "raw_path2",
        sort=False
    ):

        print("\nProcessing:", os.path.basename(raw_path))

        eeg = read_raw_eeg(raw_path)

        grouped_trials = file_meta.groupby(
            ["participant", "trial_uid"],
            sort=False
        )

        for (participant, trial_uid), trial_meta in grouped_trials:

            trial_meta = trial_meta.sort_values("window_id")

            if len(trial_meta) != 156:
                raise ValueError(
                    f"{trial_uid}: expected 156 windows, got {len(trial_meta)}"
                )

            if trial_meta["window_id"].tolist() != list(range(1, 157)):
                raise ValueError(f"Incorrect window IDs: {trial_uid}")

            # Verify metadata intervals are exactly 0.064 seconds
            starts = trial_meta["task_start_used"].to_numpy(dtype=float)

            if not np.allclose(np.diff(starts), 8 / FS, atol=1e-6):
                raise ValueError(
                    f"Unexpected time intervals: {trial_uid}"
                )

            # Calculate the starting row only once
            first_idx = int(round(starts[0] * FS))

            windows = []

            for window_number in range(156):

                # Always advance exactly 8 rows
                idx = first_idx + window_number * 8

                if idx < 0 or idx + 8 > len(eeg):
                    raise ValueError(
                        f"{trial_uid}, window {window_number + 1}: "
                        f"invalid row index {idx}"
                    )

                window = eeg[idx:idx + 8]

                if window.shape != (8, 15):
                    raise ValueError(
                        f"Incorrect window shape: {trial_uid}"
                    )

                if not np.isfinite(window).all():
                    raise ValueError(
                        f"Non-finite EEG values: {trial_uid}"
                    )

                windows.append(window.reshape(120).tolist())

            training_trials.append({
                "participant": participant,
                "trial_uid": trial_uid,
                "label": trial_meta["label"].iloc[0],
                "windows": windows
            })

    return training_trials


trials = extract_all_trials(metadata)

print("\nExtraction completed!")
print("Total extracted trials:", len(trials))
```

```python
data1 = []

for trial in trials:

    row = {
        "participant": trial["participant"],
        "trial_uid": trial["trial_uid"]
    }

    for i, window in enumerate(
        trial["windows"],
        start=1
    ):

        row[f"W{i:03d}"] = json.dumps(window)

    row["label"] = trial["label"]

    data1.append(row)


df1 = pd.DataFrame(data1)

file1 = os.path.join(
    OUTPUT_DIR,
    "training_data1.csv"
)

df1.to_csv(file1, index=False)

print("Training Data 1 saved!")
print("Shape:", df1.shape)
print("Location:", file1)
```

```python
data2 = []

for trial in trials:

    windows = trial["windows"]

    segments = [
        windows[:78],
        windows[78:]
    ]

    for segment_number, segment in enumerate(
        segments,
        start=1
    ):

        row = {
            "participant": trial["participant"],
            "trial_uid": trial["trial_uid"],
            "segment": segment_number
        }

        for i, window in enumerate(
            segment,
            start=1
        ):

            row[f"W{i:03d}"] = json.dumps(window)

        row["label"] = trial["label"]

        data2.append(row)


df2 = pd.DataFrame(data2)

file2 = os.path.join(
    OUTPUT_DIR,
    "training_data2.csv"
)

df2.to_csv(file2, index=False)

print("Training Data 2 saved!")
print("Shape:", df2.shape)
print("Location:", file2)
```
